```python
import os
import shutil
masks = os.listdir("images/masks")
dji = os.listdir("E:/Rothaut_Masterthesis/jpg")
for i in masks:
    num = i.split("_")[1].split(".")[0]
    for name in dji:
        if num in name:
            shutil.copy2(os.path.join("E:/Rothaut_Masterthesis/jpg",name), os.path.join("images/imgs",name))
```
