```python
import cv2
import numpy as np
import os

# --- Pfade anpassen ---
input_dir = "raw_masks/filled"  # Ordner mit den Eingabebildern
output_dir = os.path.join(os.path.dirname(input_dir), "bearbeitet")

# Ausgabeverzeichnis erstellen, falls nicht vorhanden
os.makedirs(output_dir, exist_ok=True)

# Farbcodes in BGR (OpenCV nutzt BGR, nicht RGB!)
BLUE = [255, 0, 0]
YELLOW = [0, 255, 255]
RED = [0, 0, 255]

# Über alle Dateien im Eingabeordner iterieren
for filename in os.listdir(input_dir):
    filepath = os.path.join(input_dir, filename)

    # Nur Bilddateien verarbeiten
    if not filename.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")):
        continue

    img = cv2.imread(filepath)

    if img is None:
        print(f"⚠️ Konnte {filename} nicht laden")
        continue

    # Prüfen, ob Blau oder Gelb im Bild vorkommt
    mask_blue = cv2.inRange(img, np.array(BLUE), np.array(BLUE))
    mask_yellow = cv2.inRange(img, np.array(YELLOW), np.array(YELLOW))

    if cv2.countNonZero(mask_blue) > 0 or cv2.countNonZero(mask_yellow) > 0:
        # Alle roten Pixel finden
        mask_red = cv2.inRange(img, np.array(RED), np.array(RED))

        # Rote Pixel auf Schwarz setzen
        img[mask_red > 0] = [0, 0, 0]

    # Ergebnis speichern
    outpath = os.path.join(output_dir, filename)
    cv2.imwrite(outpath, img)

    print(f"✅ gespeichert: {outpath}")
```
